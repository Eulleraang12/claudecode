```
# ── 2. Conectar ao Google Drive ────────────────────────────────────────────────
from google.colab import drive
drive.mount('/content/drive')
print('Drive conectado!')
```

```
# ── 3. Configuração ────────────────────────────────────────────────────────────
import os

# Caminho da pasta no seu Drive.
# Se a pasta for compartilhada, adicione um atalho:
#   Google Drive > clique direito na pasta > Organizar > Adicionar atalho ao Meu Drive
# Depois ajuste o caminho abaixo (ex: 'Meu Drive/NomeDaPasta')
PASTA_DRIVE = '/content/drive/MyDrive/TranscricaoCPL'  # ← ajuste se necessário

MODELO = 'medium'  # tiny | base | small | medium | large

EXTENSOES_VIDEO = {'.mp4', '.mkv', '.avi', '.mov', '.webm', '.flv', '.wmv', '.m4v', '.mpg', '.mpeg'}
EXTENSOES_AUDIO = {'.mp3', '.wav', '.m4a', '.ogg', '.flac', '.aac'}

# Lista os arquivos de mídia encontrados
arquivos = [
    os.path.join(PASTA_DRIVE, f)
    for f in os.listdir(PASTA_DRIVE)
    if os.path.splitext(f)[1].lower() in (EXTENSOES_VIDEO | EXTENSOES_AUDIO)
]

print(f'Pasta: {PASTA_DRIVE}')
print(f'Modelo: {MODELO}')
print(f'Arquivos encontrados: {len(arquivos)}')
for a in arquivos:
    print(f'  - {os.path.basename(a)}')
```

```
# ── 4. Transcrever e salvar na mesma pasta ─────────────────────────────────────
import whisper

if not arquivos:
    print('Nenhum vídeo/áudio encontrado. Verifique o caminho da pasta.')
else:
    print(f'Carregando modelo {MODELO}...')
    model = whisper.load_model(MODELO)
    print('Modelo carregado!\n')

    total = len(arquivos)
    for i, caminho in enumerate(arquivos, 1):
        nome = os.path.basename(caminho)
        nome_sem_ext = os.path.splitext(nome)[0]
        saida = os.path.join(PASTA_DRIVE, f'{nome_sem_ext}.txt')

        print(f'[{i}/{total}] Transcrevendo: {nome} ...')
        try:
            resultado = model.transcribe(caminho, verbose=False)
            with open(saida, 'w', encoding='utf-8') as f:
                f.write(resultado['text'].strip())
            print(f'       Salvo: {nome_sem_ext}.txt')
        except Exception as e:
            print(f'       ERRO: {e}')

    print(f'\nPronto! {total} transcrição(ões) salva(s) na pasta do Drive.')
```
